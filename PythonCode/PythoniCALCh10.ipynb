## Example of computing the area of an ellipse by a planimeter

```python
#Define parameters
n = 1000
a = 4
b = 2
t = np.linspace(0, 2*np.pi, n+1)
x = a*np.cos(t)
y = a*np.sin(t)
s = np.zeros(n)
```

```python
#Use for loop for planimeter
for i in range(n-1):
    s[i] = (1/2)*(-1*y[i]*(x[i+1]-x[i]) + x[i]*(y[i+1]-y[i]))

#Sum array to find area
A = np.sum(s)
print('Calculated of an ellipse:', A)
```

```python
#Plot graph of ellipse
plt.figure(figsize = (8,8))
plt.plot(x, y)
plt.title('Ellipse with a = 4, b = 2')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

## Exercise 10.6(a): Model flower petals

```python
#Define parameters to plot flower petal function 
a = 3
k = 4 #Determins num of petals
n = 1000
omg = 2
t = np.linspace(0, 2*np.pi, n)
theta = omg*t
r = a*np.cos(k*theta)
x = r*np.cos(theta)
y = r*np.sin(theta)
```

```python
plt.figure(figsize=(8,8))
plt.plot(x, y, color='pink', linewidth=2)
plt.title('An Eight-Petal Rose')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
#Inspect first 125 points
plt.plot(x[:125], y[:125], color='pink', linewidth=2)
plt.title('An Eight-Petal Rose')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
#Plot just one petal
plt.plot(x[32:94], y[32:94], color='red', linewidth=2)
plt.title('An Eight-Petal Rose')
plt.xlabel('x')
plt.ylabel('y')
plt.xlim(-3,3)
plt.ylim(-3,3)
plt.show()
```

```python
#Green's theorem to calculate area of petal
x_petal = x[32:94]
y_petal = y[32:94]
m = len(x_petal)
s = np.zeros(m)

for i in range(m-1):
    s[i] = x_petal[i]*(y_petal[i+1] - y_petal[i])

A = np.sum(s)
print('Area of one petal:', A)
```

```python
print('Area of entire flower:', A*8)
```
